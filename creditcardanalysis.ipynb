### Advance Analysis(Feature Engineering)


```python
df['Avg_Transaction_Value'] = df['Total_Trans_Amt']/df['Total_Trans_Ct'] 
```

```python
df.groupby('Churn')['Avg_Transaction_Value'].mean() 
```

```python
df.groupby('Income_Category')['Churn'].mean().sort_values(ascending=False) 
```

```python
def segment_transactions(x):
    if x < 40:
        return "Low Activity"
    elif x < 80:
        return "Medium Activity"
    else:
        return "High Activity"

df['Activity_Segment'] = df['Total_Trans_Ct'].apply(segment_transactions) 
```

```python
df.groupby('Activity_Segment')['Churn'].mean() 
```

```python
df.groupby('Churn')['Credit_Limit'].mean() 
```

### Risk Category

```python
def risk_category(row):
    if row['Total_Trans_Ct'] < 50 and row['Total_Trans_Amt'] < 3500:
        return "High Risk"
    elif row['Total_Trans_Ct'] < 70:
        return "Medium Risk"
    else:
        return "Low Risk"


df['Risk_Level'] = df.apply(risk_category,axis=1)  
```

```python
df.groupby('Risk_Level')['Churn'].mean().sort_values(ascending=False)  
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(x='Activity_Segment', y='Churn', data=df)  
plt.show() 
```

 Customers in the low activity segment show significantly higher churn rates, indicating that transaction frequency is a key driver of customer retention.
